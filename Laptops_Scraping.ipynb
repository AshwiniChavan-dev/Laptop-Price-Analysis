```python
def laptops_scrap(soup):

    page_data = []

    products = soup.find_all(class_="jIjQ8S")

    for device in products:

        name = None
        price = None
        rating = None
        review = None

        processor = None
        ram = None
        operating_system = None
        storage = None
        display_size = None
        microsoft_365 = None
        warranty = None

        try:
            name = device.find(
                class_="RG5Slk"
            ).get_text(" ", strip=True)
        except:
            pass

        try:
            price = device.find(
                class_="hZ3P6w DeU9vF"
            ).get_text(" ", strip=True)
        except:
            pass

        try:
            rating = device.find(
                class_="MKiFS6"
            ).get_text(" ", strip=True)
        except:
            pass

        try:
            review = (
                device.find(class_="PvbNMB")
                .find_all("span")[-1]
                .get_text(" ", strip=True)
            )
        except:
            pass

        # Specifications
        for li in device.find_all("li"):

            text = li.get_text(" ", strip=True)

            if "processor" in text.lower():
                processor = text

            elif "ram" in text.lower():
                ram = text

            elif "operating system" in text.lower():
                operating_system = text

            elif any(
                x in text.upper()
                for x in ["SSD", "HDD", "NVME"]
            ):
                storage = text

            elif (
                "display" in text.lower()
                or "inch" in text.lower()
            ):
                display_size = text

            elif any(
                x in text.lower()
                for x in [
                    "microsoft 365",
                    "microsoft office",
                    "ms office",
                    "mso"
                ]
            ):
                microsoft_365 = text

            elif "warranty" in text.lower():
                warranty = text

        page_data.append({
            "Name": name,
            "Price": price,
            "Rating": rating,
            "Total_Reviews": review,
            "Processor": processor,
            "RAM": ram,
            "Operating_System": operating_system,
            "Storage": storage,
            "Display_Size": display_size,
            "Microsoft_365_Subscription": microsoft_365,
            "Warranty": warranty
        })

    return page_data
```

```python
df = pd.DataFrame(all_data)

df.to_csv(
    "flipkart_laptops.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nScraping completed!")
print("Total records:", len(df))
print("CSV file: flipkart_laptops.csv")
```
